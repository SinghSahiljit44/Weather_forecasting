# Feature Engineering

Sometimes, original features alone are not enough, so we need to adapt and invent something new. 

In this notebook, we will talk about the feature we engineered and how and why we use them. All of these "new" features can be found inside *preprocessing.py*.

## Seasonality
As we always say in Italian, "in-between seasons don't exist anymore": this does not apply when we are data sciencing around. We added two new features:
* **doy_sin**
* **doy_cos**

These are used to compute the seasonality associated to each day. This is important, because by adding this we can add the weather's behavior to each day, so we can firmly say that on summer it is hot and on winter it is cold.

## Lags
Unlike Minecraft Bedwars, lag here is fundamental, because it keeps track of the weather's behaviour in the previous *x* days. It can be chosen how many lag days we keep, this fits well with the use of Optuna.

Do a barrel roll! They give us a smoothed summary of the last 7 or 30 days, which describes the current regime we are in at the moment.
