# Weather Forecasting
This is a project made by Sahiljit Singh and Matteo Lublanis for the Machine Learning course held by Ivan Serina and Alfonso Emilio Gerevini at Università degli Studi di Brescia. This is also one of the projects proposed by AI4WATER, a research project which aims to improve the usage of water by using AI methods.

Our goal in this project is to predict temperature and precipitations over different time horizons (days/months). We will use regression over historical sequential data and we'll also implement a simple LSTM network.
